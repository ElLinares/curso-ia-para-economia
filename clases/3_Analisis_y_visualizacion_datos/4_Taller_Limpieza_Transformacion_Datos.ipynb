# Ejercicio 9: Groupby

Agrupar los datos por categorías, nos permite realizar análisis más profundos y obtener insights valiosos, que a simple vista no vemos en la base de datos.

1. Crea un nuevo dataframe llamado "df_sexo" donde se agrupe por 'sexo' y se calcule la suma de las columnas 'poblacion_total' y 'ocupados'. Crea una nueva columna que se llame 'porcentaje_ocupados' donde dividas las columna 'ocupados' sobre 'poblacion_total'. Al final usa reset_index() para convertir las categorías de sexo en una columna.

```python
### TU CÓDIGO AQUÍ ###
df_sexo = df_empleo.groupby('sexo').agg(suma_poblacion_total=('poblacion_total', 'sum'), suma_ocupados=('ocupados', 'sum')).reset_index()
df_sexo['porcentaje_ocupados'] = df_sexo['suma_ocupados'] / df_sexo['suma_poblacion_total']
df_sexo.head()
```

2. Crea un nuevo dataframe llamado "df_educacion" donde se agrupe por 'nivel_educativo' y se calcule la suma de las columnas 'poblacion_total' y 'ocupados'. Crea una nueva columna que se llame 'porcentaje_ocupados' donde dividas las columna 'ocupados' sobre 'poblacion_total'. Al final usa reset_index() para convertir las categorías de nivel educativo en una columna.

```python
### TU CÓDIGO AQUÍ ###
df_educacion =df_empleo.groupby('nivel_educativo').agg(
    suma_poblacion_total=('poblacion_total','sum'), suma_ocupados=('ocupados','sum')).reset_index()
df_educacion['porcentaje_ocupados'] = df_educacion['suma_ocupados'] / df_educacion['suma_poblacion_total']
df_educacion.head()
```

# Ejercicio 10: Filtrar datos

1. A partir de df_empleo, crea el DataFrame `df_ing` con observaciones donde `ingreso_laboral` no sea nulo y además sea mayor estrictamente al salario mínimo para 2026 (con subsidio de transporte).

```python
### TU CÓDIGO AQUÍ ###
salario_minimo_2026 = 2000000

df_ing = df_empleo[
    (df_empleo['ingreso_laboral'].notna()) &
    (df_empleo['ingreso_laboral'] > salario_minimo_2026)
].reset_index(drop=True)

print(df_ing.shape)
df_ing
```

2. ¿Cuántas personas tienen un salario estrictamente mayor al salario mínimo?

```python
### TU CÓDIGO AQUÍ ###
df_ing[df_ing['ingreso_laboral']>salario_minimo_2026]
print(df_ing.shape)
```
